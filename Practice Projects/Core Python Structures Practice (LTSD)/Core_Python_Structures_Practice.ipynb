```python
temp=[]
def rmvdup (lst):
    for i in lst:
        if i not in temp:
            temp.append(i)
        else:
            lst.remove(i)
    temp.sort()
    return temp
```

```python

def lcompre (lst):
    lst.clear()
    for i in range(1,21,2):
        lst.append(i**2)
    return lst
```

```python
def largsmal (lst):
    small=min(lst)
    larg=max(lst)
    print(f"Smallest is {small}\n Largest is {larg}")
```

# Mini Chalange

Create a summary report using all 4 data structures: (List, Tuple, Set, Dictionary)
Use a list to hold multiple student records.
Each record should be a tuple of (name + subject-wise marks).
Use a set to track unique subjects.
Use a dictionary inside each tuple to store subject-wise marks.

```python
def student_Entry():
    student_record=list()
    no_of_stu=int(input("Enter the number of student: "))
    for i in range(no_of_stu):
        name=input("Enter Student Name:")
        subject_wise_marks=dict()
        no_of_sub=int(input(f"Enter the Number of Subject for {name} "))
        for j in range(no_of_sub):
            subject=input("Enter Subject Name:")
            score=int(input(f"Enter Marks in {subject}:"))
            subject_wise_marks[subject]=score
        student_record.append((name,subject_wise_marks))
    return student_record

def output(x):
    all_subject=set()
    for name,marks in x:
        total=sum(marks.values())
        avg=total/len(marks)
        all_subject.update(marks.keys())
        print(f"{name}: Average= {avg}")
    print(all_subject)


def main():
    students=student_Entry()
    output(students)
    print("------------")
    print(students)
main()





```
