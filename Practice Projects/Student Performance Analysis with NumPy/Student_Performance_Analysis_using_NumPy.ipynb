```python
import numpy as np
```

# You create a 2D array that stores marks for a number of students across different subjects.

Example:
Rows = Students,
Columns = Subjects,
So a 50×5 array means: 50 students,
5 subjects (e.g., Math, Physics, Chemistry, Biology, English)

```python
# Generating Random Marks for Students
marks=np.random.randint(40,100, size=(50,5))
subjects=["Math", "Physics", "Chemistry", "Biology", "English"]
print(marks)
```

# Performing Statistical Analysis
1. Average Marks per Student
2. Average Marks per Subject
3. Maximum and Minimum Marks per Subject
4. Standard Deviation for Each Subject

```python
# 1. Average Marks per Student
avg_marks_per_student=np.mean(marks, axis=1)
print(avg_marks_per_student.reshape(50,1))
```

```python
# 2. Average Marks per Subject
avg_marks_per_subject=np.mean(marks, axis=0)
avg_marks_per_subject.reshape(1,5)
print(avg_marks_per_subject)
```

```python
# 3. Maximum and Minimum Marks Per Subject

maximum=np.max(marks, axis=0)
minimum=np.min(marks, axis=0)
print("Maximum marks in each subject: ",maximum)
print("Minimum marks in each subject: ",minimum)
```

```python
# 4. Standard Daviation for Each Subject
std_each_subject=np.std(marks, axis=0)
print("Standard Daviation for Each Subject: ",std_each_subject)
```

# Identify Top and Bottom Students

What you're doing:
1. Find the student who has the highest overall average.
2. Find the student with the lowest overall average.

```python
avg=np.mean(marks, axis=1)
avg.reshape(50,1)
top=np.argmax(avg)
print("Student with Highest Overall Score: ",top)
bottom=np.argmin(avg)
print("Student with Lowest Overall Score: ",bottom)
```

# Identify Toppers per Subject

For each subject, find the student who got the highest score.

Then print:
1. Subject name
2. Student index
3. Mark obtained

```python
subject_topper=np.argmax(marks, axis=0)
print(subject_topper)

for i,sub in enumerate(subjects):
    student_index=subject_topper[i]
    subject_name=sub
    mark_obtain=marks[subject_topper[i],i]
    print(f"Topper in {sub}: Student {student_index} with {mark_obtain} marks")
```
