```python
def Add_Student_Record():
    student_marks=dict()
    name=input("Enter Student Name: ")
    Math=int(input("Enter Obtained Score in Math: "))
    Science=int(input("Enter Obtained Score in Science: "))
    English=int(input("Enter Obtained Score in English: "))
    line=f"{name}, Math: {Math}, Science: {Science}, English: {English}\n"

    with open('students.txt', mode='a') as file:
        file.write(line)
        print("Student Data Recorded Successfully")
```

```python
def View_All_Records():
    with open('students.txt', mode='r') as file:
        lines=file.readlines()
    for i in lines:
        print(i.strip())
```

```python
def Search_Student_by_Name():
    flag=0
    s_name=input("Enter the student you are looking for: ")
    with open('students.txt', mode='r') as file:
        lines=file.readlines()
    for i in lines:
        lst=i.strip().split(",")
        if lst[0].title()==s_name.title():
            flag=1
            print("Student Details: ")
            print(i)
            break
    if flag==0:
        print("Student not Found.")
```

```python
def Delete_Student_Record():
    filtered=[]
    found=False
    s_name=input("Enter the student you want delete.")
    with open('students.txt', mode='r') as file:
        lines=file.readlines()
    for i in lines:
        lst=i.strip().split(",")
        if lst[0].strip().lower()!=s_name.lower():
            filtered.append(i)
        else:
            found=True
            continue
    with open('students.txt', mode='w') as dest_file:
            dest_file.writelines(filtered)
    if not found:
        print("The Student is Not Found")
    else:
        print("The Student Data Deleted Successfully")
```

```python
def Update_Marks():
    s_name=input("Enter a student name you want to update his details.")
    updated_lines=[]
    while(True):
        updated_marks=input(f"Enter updated marks of {s_name}.")
        updated_marks=updated_marks.strip().split(",")
        if len(updated_marks) != 3:
            print("Please provide exactly 3 comma-separated marks (e.g. 80,85,90).")
        else:
            break
    with open('students.txt', mode='r') as file:
        lines=file.readlines()
    for i in lines:
        lst=i.strip().split(",")
        if lst[0].strip().lower()!=s_name.lower():
            updated_lines.append(i)
        else:
            line=f"{lst[0]}, Math: {updated_marks[0]}, Science: {updated_marks[1]}, English: {updated_marks[2]}\n"
            updated_lines.append(line)
            print("Marks Updated Successfully")
    with open('students.txt', mode='w') as file:
        file.writelines(updated_lines)
```

```python
def clear_file():
    with open('students.txt', mode='w') as file:
        pass
```

```python
def main():
    print("****Student Report Card Manager****")
    print("1. Add Student Record")
    print("2. View All Records")
    print("3. Search Student by Name")
    print("4. Delete Student Record")
    print("5. Update Marks")
    print("6. Exit")
    while(True):
        choice=int(input("Enter Your Choice: "))
        if choice==1:
            Add_Student_Record()
        elif choice==2:
            View_All_Records()
        elif choice==3:
            Search_Student_by_Name()
        elif choice==4:
            Delete_Student_Record()
        elif choice==5:
            Update_Marks()
        elif choice==6:
            print("Code Successfully Executed.")
            clear_file()
            break
if __name__=="__main__":
    main()
```
