```python
import numpy as np
import random as rd
```

# Create the Environment
1. Simulate a 4-lane intersection (North, South, East, West).
2. Each lane has a queue (initially filled with 0–20 cars randomly).
3. Represent the traffic state using a NumPy array, e.g.,

```python
lane_direction=["North", "South", "East", "West"]
intersection=np.random.randint(0,20, size=(4))
print(intersection)
```

# Simulate Light Timing
1. Green light allows 1 car to pass every second.
2. Red light allows no cars to pass.
3. Simulate a full cycle: NS green for X seconds, then EW green for Y seconds.
4. Subtract cars from the lanes depending on green/red duration.
5. Add random new cars (0–5) to each lane every cycle.

```python
green_ns=rd.randint(5,10)
green_ew=rd.randint(5,10)
intersection[0]=np.maximum(0, intersection[0] - green_ns)
intersection[1]=np.maximum(0, intersection[1] - green_ns)
intersection[2]=np.maximum(0, intersection[2] - green_ew)
intersection[3]=np.maximum(0, intersection[3] - green_ew)
new_cars=np.random.randint(0,6, size=(4))
intersection+=new_cars
for ind,i in enumerate(lane_direction):
    direction=lane_direction[ind]
    no_of_car=intersection[ind]
    print(f"{direction}: {no_of_car}")
```

# Run the Simulation
1. Run for 20–50 cycles.
2. For each cycle, update the car counts.
3. Track metrics:
    - Average waiting time.
	- Total cars passed.
	- Congestion (queue length).


```python
run=rd.randint(20,51)
avg_waiting_list=[]
congestion_list=[]
total_car_passed=np.array([0,0,0,0])
for i in range(run):
    green_ns=rd.randint(5,10)
    green_ew=rd.randint(5,10)
    before=intersection.copy()
    intersection[0]=np.maximum(0, intersection[0] - green_ns)
    intersection[1]=np.maximum(0, intersection[1] - green_ns)
    intersection[2]=np.maximum(0, intersection[2] - green_ew)
    intersection[3]=np.maximum(0, intersection[3] - green_ew)
    car_passed=(before-intersection)
    total_car_passed+=car_passed
    new_cars=np.random.randint(0,6, size=(4))
    intersection+=new_cars
    congestion=np.sum(intersection)
    congestion_list.append(congestion)
    avg_waiting=np.mean(intersection)
    avg_waiting_list.append(avg_waiting)
avg_congestion=np.mean(congestion_list)
final_waiting=np.mean(avg_waiting_list)
print("Total Number of Cycle: ",run)
print(f"Average Waiting Time: {final_waiting: .2f}")
print("Total Car Passed through Each Lane: ",total_car_passed)
print(f"Average congestion in the lanes: {avg_congestion: .2f}")
```

# Optimization
1. Try different combinations of green light duration (X and Y).
2. Find the combination that minimizes average queue length or maximizes flow.
3. Use NumPy to iterate over multiple combinations and track best results.

```python
run=35
green_light_list=[]
optimizer=np.zeros((35,3))
for i in range(run):
    green_ns=rd.randint(5,11) #fact 4
    green_ew=rd.randint(5,11) #fact 4
    before=intersection.copy()
    intersection[0]=np.maximum(0, intersection[0] - green_ns)
    intersection[1]=np.maximum(0, intersection[1] - green_ns)
    intersection[2]=np.maximum(0, intersection[2] - green_ew)
    intersection[3]=np.maximum(0, intersection[3] - green_ew)
    car_passed=np.sum(before-intersection) #fact 1
    new_cars=np.random.randint(0,6, size=(4))
    intersection+=new_cars
    congestion=np.sum(intersection) #fact 2
    avg_waiting=np.mean(intersection) #fact 3

    optimizer[i]= [car_passed, congestion, avg_waiting]
    green_light_list.append((green_ns,green_ew))

indx_max_car=np.argmax(optimizer[0:, 0])
print(f"Max Car Passed: {optimizer[indx_max_car, 0]} at green pair {green_light_list[indx_max_car]}")

indx_min_congestion=np.argmin(optimizer[0:, 1])
print(f"Min Congestion: {optimizer[indx_min_congestion, 1]} at green pair {green_light_list[indx_min_congestion]}")

indx_min_avg_wtime=np.argmin(optimizer[0:, 2])
print(f"Min Avg Waiting: {optimizer[indx_min_avg_wtime, 2]:.2f} at green pair {green_light_list[indx_min_avg_wtime]}")


    
```

```python
#3. Use NumPy to iterate over multiple combinations and track best results.
green_light_list=[]
optimizer=np.zeros((36,3))
n=0
for i in range(5,11,1):
    for j in range(5,11,1):
        green_ns=i #fact 4
        green_ew=j #fact 4
        before=intersection.copy()
        intersection[0]=np.maximum(0, intersection[0] - green_ns)
        intersection[1]=np.maximum(0, intersection[1] - green_ns)
        intersection[2]=np.maximum(0, intersection[2] - green_ew)
        intersection[3]=np.maximum(0, intersection[3] - green_ew)
        car_passed=np.sum(before-intersection) #fact 1
        new_cars=np.random.randint(0,6, size=(4))
        intersection+=new_cars
        congestion=np.sum(intersection) #fact 2
        avg_waiting=np.mean(intersection) #fact 3

        optimizer[n]= [car_passed, congestion, avg_waiting]
        n+=1
        green_light_list.append((green_ns,green_ew))

indx_max_car=np.argmax(optimizer[0:, 0])
print(f"Max Car Passed: {optimizer[indx_max_car, 0]} at green pair {green_light_list[indx_max_car]}")

indx_min_congestion=np.argmin(optimizer[0:, 1])
print(f"Min Congestion: {optimizer[indx_min_congestion, 1]} at green pair {green_light_list[indx_min_congestion]}")

indx_min_avg_wtime=np.argmin(optimizer[0:, 2])
print(f"Min Avg Waiting: {optimizer[indx_min_avg_wtime, 2]:.2f} at green pair {green_light_list[indx_min_avg_wtime]}")
arr=np.array([[green_light_list[indx_max_car][0],green_light_list[indx_max_car][1]], [green_light_list[indx_min_congestion][0],green_light_list[indx_min_congestion][1]], [green_light_list[indx_min_avg_wtime][0],green_light_list[indx_min_avg_wtime][1]]])

mean=np.mean(arr, axis=0)
mean=np.round(mean)
print(f"Iterated over multiple combinations to get best result: {mean}")

```
